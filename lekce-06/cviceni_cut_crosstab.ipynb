# Cvičení — `pd.cut` a `pd.crosstab` na různých datasetech

Procvičíš si binning (`pd.cut`) a kontingenční tabulky (`pd.crosstab`) na pěti různých datasetech, které jsou zabudované přímo v seabornu — žádné stahování souborů.

**Společný setup pro všechny úlohy:**

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

Každý dataset načteš jedním řádkem: `df = sns.load_dataset('jméno_datasetu')`.

## 1. 🍽️ Tips — spropitné v restauraci

**Načtení:**
```python
tips = sns.load_dataset('tips')
tips.head()
```

**Sloupce:** `total_bill` (cena účtu v $), `tip` (spropitné v $), `sex`, `smoker` (kuřák? Yes/No), `day` (den v týdnu), `time` (Lunch/Dinner), `size` (počet lidí u stolu).

### Zadání

**A) Binning ceny účtu (`pd.cut`)**
Rozděl `total_bill` do tří kategorií podle vlastních hranic: `levný účet` (do 15 $), `střední účet` (15–30 $), `drahý účet` (nad 30 $). Výsledek ulož do nového sloupce `bill_category`.

**B) Procento spropitného + binning**
Spočítej procento spropitného z účtu (`tip / total_bill * 100`) a rozděl ho do kategorií: `skoupý` (do 10 %), `standard` (10–20 %), `štědrý` (nad 20 %). Nový sloupec pojmenuj `tip_pct_category`.

**C) Crosstab — počty**
Spočítej, kolik lidí v které kombinaci `tip_pct_category` × `time` (oběd vs. večeře) dalo spropitné. Přidej součty řádků i sloupců (`margins=True`).

**D) Crosstab — agregace**
Spočítej **průměrné spropitné v $** podle `bill_category` (řádky) a `day` (sloupce). Použij `values=` a `aggfunc='mean'`.

**E) Otázka k zamyšlení**
Z výsledku D) — dávají větší spropitné v $ lidé z drahých účtů? A v procentech (z C)?

## 2. 🐧 Penguins — tučňáci na Antarktidě

**Načtení:**
```python
penguins = sns.load_dataset('penguins')
penguins.head()
```

**Sloupce:** `species` (Adelie / Chinstrap / Gentoo), `island` (ostrov), `bill_length_mm` (délka zobáku), `bill_depth_mm` (hloubka zobáku), `flipper_length_mm` (délka ploutví), `body_mass_g` (váha v gramech), `sex`.

*Pozor:* dataset obsahuje NaN hodnoty — `cut` si s nimi poradí (nechá je NaN), ale ber to v potaz.

### Zadání

**A) Binning váhy (`pd.cut`)**
Rozděl `body_mass_g` do čtyř kategorií: `velmi malý` (do 3500 g), `malý` (3500–4500 g), `velký` (4500–5500 g), `velmi velký` (nad 5500 g). Sloupec pojmenuj `size_category`.

**B) Crosstab — kolik kterých druhů**
Vytvoř kontingenční tabulku `species` × `island` — kolik tučňáků každého druhu žije na jakém ostrově. Použij `margins=True`.

**C) Crosstab s procenty**
Spočítej tu samou tabulku, ale s `normalize='columns'` — kolik **procent** tučňáků na každém ostrově patří k jakému druhu. Co ti to říká o struktuře ostrovů?

**D) Crosstab — agregace**
Spočítej **průměrnou délku ploutví** (`flipper_length_mm`) podle druhu (řádky) a pohlaví (sloupce).

**E) Spojení cut + crosstab**
Zobraz kontingenční tabulku `size_category` × `species` — kolik tučňáků každé velikostní kategorie patří k jakému druhu. Najdeš druh, který má jasně dominantní velikost?

## 3. 🚗 MPG — spotřeba aut

**Načtení:**
```python
mpg = sns.load_dataset('mpg')
mpg.head()
```

**Sloupce:** `mpg` (míle na galon — pozor, **vyšší = úspornější**), `cylinders` (počet válců), `displacement` (objem motoru), `horsepower` (výkon v koních), `weight` (váha v librách), `model_year` (rok 70–82), `origin` (usa / japan / europe), `name`.

### Zadání

**A) Binning úspornosti (`pd.cut`)**
Rozděl `mpg` do tří kategorií: `žrout` (do 18), `průměr` (18–28), `úsporné` (nad 28). Sloupec pojmenuj `economy`.

**B) Binning éry (`pd.cut`)**
Rozděl `model_year` do tří dekádních skupin: `začátek 70s` (70–73), `polovina 70s` (74–77), `konec 70s a 80s` (78–82). Sloupec pojmenuj `era`. *Tip: `pd.cut` s `bins=[69, 73, 77, 82]` — všimni si, že dolní hranice musí být menší než nejnižší hodnota.*

**C) Crosstab — úspornost vs. původ**
Vytvoř kontingenční tabulku `economy` × `origin`. Použij `normalize='columns'`, abys viděla procenta. Která země vyráběla nejúspornější auta?

**D) Crosstab — agregace**
Spočítej **průměrný výkon** (`horsepower`) podle `era` (řádky) a `origin` (sloupce). Klesal výkon v čase ve všech zemích stejně?

**E) Bonus**
Pomocí crosstabu zjisti **medián váhy** (`weight`) podle `cylinders` a `origin`. *Tip: `aggfunc='median'`.*

## 4. 💎 Diamonds — ceny a kvalita diamantů

**Načtení:**
```python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
```

**Sloupce:** `carat` (váha v karátech), `cut` (kvalita brusu: Fair / Good / Very Good / Premium / Ideal), `color` (D nejlepší → J nejhorší), `clarity` (čistota: I1, SI2, SI1, VS2, VS1, VVS2, VVS1, IF), `depth`, `table`, `price` ($), `x`, `y`, `z` (rozměry mm).

*Velký dataset (~54 000 řádků) — všechno poběží v pohodě, ale výpisy `head()` ukazují jen špičku.*

### Zadání

**A) Binning ceny (`pd.cut`)**
Rozděl `price` do čtyř kategorií podle vlastních hranic v dolarech: `dostupné` (do 1000), `střední třída` (1000–5000), `prémium` (5000–15000), `luxus` (nad 15000). Sloupec pojmenuj `price_tier`.

**B) Binning karátů (`pd.cut`)**
Rozděl `carat` do kategorií: `malý` (do 0.5), `střední` (0.5–1.0), `velký` (1.0–2.0), `obří` (nad 2.0). Sloupec pojmenuj `carat_size`.

**C) Crosstab — kvalita vs. cenová třída**
Vytvoř kontingenční tabulku `cut` (řádky) × `price_tier` (sloupce). Použij `normalize='index'` — kolik procent diamantů každé kvality brusu spadá do které cenové třídy.

**D) Crosstab — agregace**
Spočítej **průměrnou cenu** podle `carat_size` (řádky) a `cut` (sloupce). Roste cena s kvalitou brusu uvnitř každé velikostní kategorie?

**E) Otázka k zamyšlení**
V tabulce z D) porovnej cenu *malého Ideal* a *velkého Fair*. Co je dražší? Co je z toho silnější faktor — velikost, nebo brus?

## 5. 🚕 Taxis — jízdy newyorských taxíků

**Načtení:**
```python
taxis = sns.load_dataset('taxis')
taxis.head()
```

**Sloupce:** `pickup` / `dropoff` (datetime), `passengers` (počet pasažérů), `distance` (vzdálenost v mílích), `fare` (jízdné $), `tip` ($), `tolls` ($), `total` ($), `color` (yellow / green), `payment` (credit card / cash), `pickup_zone`, `dropoff_zone`, `pickup_borough`, `dropoff_borough`.

### Zadání

**A) Binning vzdálenosti (`pd.cut`)**
Rozděl `distance` do kategorií: `krátká` (do 1 míle), `střední` (1–3 míle), `dlouhá` (3–10 mil), `velmi dlouhá` (nad 10 mil). Sloupec pojmenuj `distance_category`.

**B) Binning ceny (`pd.cut`)**
Rozděl `total` do tří kategorií: `levná jízda` (do 10 $), `střední jízda` (10–25 $), `drahá jízda` (nad 25 $). Sloupec pojmenuj `price_category`.

**C) Crosstab — vzdálenost vs. cena**
Vytvoř kontingenční tabulku `distance_category` × `price_category`. Bude diagonála silná? (Měla by — krátké jízdy bývají levné, dlouhé drahé.)

**D) Crosstab — způsob platby**
Spočítej procentuální zastoupení způsobů platby (`payment`) podle vzdálenosti (`distance_category`). Použij `normalize='index'`. Platí lidé za delší jízdy spíš kartou?

**E) Crosstab — agregace**
Spočítej **průměrné spropitné** (`tip`) podle `pickup_borough` (řádky) a `payment` (sloupce). *Pozor: hotovostní spropitné se v datech většinou nezaznamenává — uvidíš v tabulce zajímavý vzor.*

## 💡 Tipy ke všem cvičením

- **`pd.cut` má o jeden label méně než hranic** — pokud `bins=[0, 10, 20, 30]`, pak `labels` má 3 prvky.
- **Hranice jsou zleva otevřené, zprava uzavřené** (`(0, 10]`). Hodnota 10 spadne do prvního intervalu, 11 do druhého.
- **Hodnoty mimo rozsah** se stanou `NaN`. Když si nejsi jistá rozsahem, použij `bins=[-np.inf, 10, 20, np.inf]`.
- **`crosstab` s `normalize=`** — `'index'` sčítá řádky na 1, `'columns'` sloupce, `'all'` celou tabulku.
- **`crosstab` umí agregovat** — pošli `values=df['sloupec']` a `aggfunc='mean'` (nebo `'median'`, `'sum'`, `'max'`...).
- **Když je výsledek vědecký zápis**, použij `.round(2)` pro hezčí výpis.
